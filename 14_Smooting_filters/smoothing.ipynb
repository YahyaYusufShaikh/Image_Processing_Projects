```python

import cv2
import numpy
```

```python
img = cv2.imread('./images/beach.jpg')

cv2.imshow('image', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Averaging

```python
blur_avg3 = cv2.blur(img, (3,3))
blur_avg5 = cv2.blur(img, (5,5))
blur_avg6 = cv2.blur(img, (6,6))    
blur_avg7 = cv2.blur(img, (7,7))
blur_avg9 = cv2.blur(img, (9,9))
```

```python
cv2.imshow('blur_avg3', blur_avg3)
cv2.imshow('blur_avg5', blur_avg5)
cv2.imshow('blur_avg6', blur_avg6)
cv2.imshow('blur_avg7', blur_avg7)
cv2.imshow('blur_avg9', blur_avg9)
cv2.imshow('image', img)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

Guassion Blug

```python
mb = cv2.imread('./images/city.jpg')

cv2.imshow('image', mb)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
blur = cv2.medianBlur(mb, 3)

cv2.imshow('blur', blur)
cv2.imshow('image', mb)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
