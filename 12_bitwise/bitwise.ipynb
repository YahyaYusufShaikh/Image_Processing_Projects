```python
import numpy as np
import cv2
```

```python
rectangle = np.zeros((300, 300), dtype='uint8')
cv2.rectangle(rectangle, (25, 25), (275, 275), 255, -1)

cv2.imshow('rectange', rectangle)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
circle = np.zeros((300, 300), dtype='uint8')
cv2.circle(circle,(150, 150), 150, 255, -1)

cv2.imshow('circle', circle)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

AND

```python
img_and = cv2.bitwise_and(rectangle, circle)

cv2.imshow('rectange', rectangle)
cv2.imshow('circle', circle)
cv2.imshow('and', img_and)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Bitwise OR

```python
img_or = cv2.bitwise_or(rectangle, circle)

cv2.imshow("Bit OR", img_or)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

Bitwise NOT

```python
img_not_rec = cv2.bitwise_not(rectangle)
img_not_cir = cv2.bitwise_not(circle)

cv2.imshow("Not Rectangle", img_not_rec)
cv2.imshow("rectange", rectangle)
cv2.imshow("circle", circle)
cv2.imshow("Not Circle", img_not_cir)
cv2.waitKey(0)
cv2.destroyAllWindows(
    
)
```

Bitwise XOR

```python
img_xor = cv2.bitwise_xor(rectangle, circle)
cv2.imshow('bit XOR', img_xor)
cv2.imshow('rectangle', rectangle)
cv2.imshow('circle', circle)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
